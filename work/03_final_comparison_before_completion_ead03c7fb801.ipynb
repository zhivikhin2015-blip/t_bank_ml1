# Итоговое сравнение ручного и автоматического подходов

## Шаг 1. Фиксируем выбранные модели

Ручной и автоматический подходы уже разработаны и проверены на validation.
Здесь начинаем подготовку к независимой итоговой проверке.

- Ручная модель использует пять признаков и выбранные настройки 200/7.
- Автоматическая модель использует пять ручных и пять отобранных новых признаков, настройки 100/7.

Настройки и наборы признаков берём из сохранённых результатов. После просмотра test не подбираем новые варианты по его качеству.

**На этом шаге читаем только метаданные. Сделки test не загружаем и его метрики не считаем.**


### Запусти следующую ячейку через Shift + Enter

Код открывает два файла с условиями экспериментов, сравнивает клиентов, активы, периоды и правила оценки. `assert` остановит выполнение, если условия различаются.

Это нужно, чтобы сравнивать модели на одинаковой задаче. Ожидаемый вывод: пять признаков у ручной модели, десять у автоматической и сообщение «Условия сравнения совпадают».


```python
from pathlib import Path
import json

project_dir = Path.cwd()
if project_dir.name == "outputs":
    project_dir = project_dir.parent

manual_record = json.loads(
    (project_dir / "outputs/manual_baseline_reviewed_results.json")
    .read_text(encoding="utf-8")
)

automatic_record = json.loads(
    (project_dir / "outputs/automatic_features_reviewed_results.json")
    .read_text(encoding="utf-8")
)

# Проверяем одинаковые условия экспериментов
for key in [
    "client_ids", "asset_ids",
    "train_months", "validation_months", "test_months",
    "transactions_sha256", "label_definition",
    "metric_definitions", "tie_break", "observation"
]:
    assert manual_record[key] == automatic_record[key], key

print("Ручная модель:")
print("Признаков:", len(manual_record["features"]))
print("Настройки:", manual_record["parameters"])

print("\nАвтоматическая модель:")
print("Признаков:", len(automatic_record["features"]))
print("Настройки:", automatic_record["parameters"])

print("\nУсловия сравнения совпадают.")
```
